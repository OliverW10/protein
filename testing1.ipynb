```python
import random
from gym_lattice.envs import Lattice2DEnv
from stable_baselines3 import DQN, PPO
import math

# HHHH x
# HHHP
# HHPH x
# HHPP
# HPHH x
# HPHP
# HPPH x
# HPPP
# PHHH
# PHHP
# PHPH
# PHPP
# PPHH
# PPHP
# PPPH
# PPPP
# 2/4 * 1/4 * 4/16

def random_seq_fn(length):
    return ''.join(random.choices('HP', k=length))

length = 20
ratio = 0.7
hs = math.floor(ratio * length)
ps = length - hs
seq = list('H' * hs + 'P' * ps)

def shuffle_seq_fn():
    random.shuffle(seq)
    return ''.join(seq)

env = Lattice2DEnv(
    20,
    shuffle_seq_fn,
    end_ceneterd=True,
    distance_penalty=0.01,
    collision_penalty=-0.5,
    render_crop=0.5
)
```

```python

model_ppo = PPO(
    "MlpPolicy",
    env,
    learning_rate=0.0001,
    gamma=1.0,
    # policy_kwargs={"net_arch": [100, 10]},
    tensorboard_log="logs",
    ent_coef=0.001
)
```

```python
model_ppo.learn(total_timesteps=500_000, tb_log_name="PPO")
```

```python
model_dqn = DQN(
    "MlpPolicy",
    env,
    tensorboard_log="logs",
    exploration_fraction=0.5,
    exploration_initial_eps=0.5,
    buffer_size=200_000
    # policy_kwargs={"net_arch": [100, 10]}
)
```

```python
model_dqn.learn(total_timesteps=500_000, tb_log_name="DQN")
```

```python
print(env.grid_length)
```

```python
# loaded_model = PPO.load("17_decent.model")

model = model_dqn
# print(obs.shape)
def run_with_seq(seq: str):
    env.seq_fn = lambda: seq
    obs, _ = env.reset()
    print("Solving for:", env.seq)
    env.render()
    while not env.done:
        # Random agent samples from action space
        action, _ = model.predict(obs)
        obs, reward, terminated, truncated, info = env.step(int(action))
        env.render()
        if terminated or truncated:
            print("Episode finished! Reward: {} | Collisions: {} | Actions: {}".format(reward, info['collisions'], info['actions']))
            break

run_with_seq("hhppppphhppphppphp".upper())

# for i_episodes in range(3):
#     run_with_seq(random_seq_fn(10))
```
